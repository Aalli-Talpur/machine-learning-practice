## Exploratory Data Visualization

This section uses visualizations to explore how the numerical and categorical features relate to the target variable, `Price`. Regression plots are used for numerical features, while average-price bar charts are used for categorical features. A correlation heatmap is also included to summarize the linear relationships between the numerical variables and the target.

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Create regression plots for numerical features against Price
num_col = ['Year','Engine Size','Mileage']
for column in num_col:
    sns.regplot(x=column, y='Price', data=data)
    plt.xlabel(column)
    plt.ylabel("Price")
    plt.show()
```

```python
# Compare average price across categorical groups
correlation_heatmap = ['Year','Engine Size','Mileage', 'Price']
sns.heatmap(data[correlation_heatmap].corr(), annot=True, fmt=".3f", cmap='coolwarm')
plt.show()
```

```python
# Display correlations between numerical features and Price
str_col = ['Brand', 'Fuel Type', 'Transmission', 'Condition', 'Model']
for column in str_col:
    string_groupby = data.groupby(column)['Price'].mean()
    string_groupby.plot(kind="bar")
    plt.xlabel(column)
    plt.ylabel("Price")
    plt.title(f"Average price by {column}")
    plt.show()
```

### Interpretation

The visualizations did not reveal any strong individual relationship between the numerical or categorical features and the target variable, `Price`. The numerical features showed almost no linear correlation with price, while the average prices across most categorical groups were relatively similar.

However, these results are not enough to conclude that the features are useless. Their predictive value may become clearer when multiple features are combined within a machine learning model. The next stage will therefore test different feature combinations through model training and evaluation.

### Conclusion

These findings suggest that additional basic visualizations may provide limited new insight. However, this does not prove that the features have no predictive value. The next stage will therefore evaluate whether combinations of numerical and categorical features can predict car prices when used together in machine learning models.
