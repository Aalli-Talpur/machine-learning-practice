# Data Visualization Practice — World Happiness Report 2019

## Task Overview

The purpose of this notebook is to improve practical data-visualization skills using the **World Happiness Report 2019** dataset.

The focus is not on machine-learning modelling. Instead, this notebook is designed to practice:

- selecting the correct chart for a specific analytical question,
- customizing plots for clarity and readability,
- comparing distributions and categories,
- exploring relationships between numerical features,
- identifying patterns and outliers,
- and writing short interpretations that explain what each visualization actually shows.

The main goal is to move beyond creating plots simply because they are part of EDA, and instead use each visualization to answer a clear question and communicate a meaningful story from the data.
