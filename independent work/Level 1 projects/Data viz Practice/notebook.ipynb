## Data import and investigate 

```python
import  pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df = pd.read_csv('2019.csv')
```

```python
df.head()
```

```python
df.describe()
```

```python
df.info()
```

```python
plt.figure(figsize=(20,9))
ax = sns.scatterplot(data=df, x='GDP per capita', y='Score', s=90, alpha=1, edgecolor='black', linewidth=2, color='slateblue')
ax.set_title("Relation between score and GDP per capita", fontsize=15, fontweight='bold',pad=18)
ax.set_xlabel('GDP Per capita', fontsize=13, fontweight='bold')
ax.set_ylabel('score', fontsize=13, fontweight='bold')
ax.set_facecolor('snow')
ax.tick_params(axis='both', labelsize=10, colors='black')
ax.grid(True, linewidth=0.9, linestyle='--', alpha=0.5)
plt.show()
```

----
## bar chart


```python
top10 = df.sort_values(by='Score', ascending=False).head(10)
```

```python
plt.figure(figsize=(20,9))
ax = sns.barplot(data=top10,x='Country or region', y='Score', hue="Country or region", palette="Spectral", legend=False, edgecolor='black', linewidth=1.5, alpha=1)
ax.set_title("Finland Leads the World's Happiest Countries in 2019", fontsize=15, fontweight='bold', pad=15)
ax.set_xlabel("Country", fontsize=12, fontweight='bold')
ax.set_ylabel("score", fontsize=12, fontweight='bold')
ax.tick_params(axis='both', labelsize=10, colors='black')
ax.grid(axis='y', linestyle='--', linewidth=1.5, alpha=0.5)
ax.set_axisbelow(True)
ax.spines['top'].set_visible(False)
ax.spines['bottom'].set_visible(False)
plt.xticks(rotation=30)
for container in ax.containers:
    ax.bar_label(container, fmt='%.2f', padding=3, fontsize=9)
plt.show()
```
