## Country Table Profile

```python
country_profile = pd.DataFrame({
    "columns":country_df.columns,
    "Data_type": country_df.dtypes.values,
    "missing_value": country_df.isna().sum().values,
    "missing_percentage": (country_df.isna().mean()*100).values,
    "unique_value":country_df.nunique().values
})
```

```python
country_profile
```

```python
country_df.shape
```

```python
country_df.duplicated().sum()
```

```python
country_df
```

### Evaluation 

The Country table is small, containing 11 records and 2 columns. Both columns are complete, with no missing values, and there are no duplicate rows. Each column contains 11 unique values, indicating that both the country IDs and country names are unique within this table. Overall, the table appears structurally clean based on these basic profiling checks.
